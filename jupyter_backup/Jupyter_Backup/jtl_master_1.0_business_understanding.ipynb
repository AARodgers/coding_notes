#### What are the specific business objectives of this project? Define as precisely as possible.

```python
Vesta Corporation, in partnership with IEEE Computational Intelligence Society, is seeking the best solutions for fraud prevention. With higher
accuracy fraud detection, customers are hassled less. Accurate fraud detection also has the potential to save millions of dollars. 

The business objective of this project is to improve the efficiency of fraudulent transaction alerts. 
```

#### What data are available to pursue those objectives?

```python
Data is available from Vesta's e-commerce transactions. There is a transaction file and an identity file, joined by TransactionID.
```

#### What are the success criteria for each of the project's objectives? Define as precisely as possible.

```python
1. Improving the efficacy of fraudulant transaction alerts for millions of people. 
2. Increase the revenue for hundreds of thousands of businesses by reducing their fraud loss. 
3. Reducing false positives. 
```

#### Describe the data mining problem type (regression, classification, clustering, etc.)

```python
It is a binary classification problem. For each transaction, we're determining whether it is or isn't fraudulant. 
```

#### What are the specific technical goals for the project?¶

```python
The goal is to have the lowest area under the ROC curve between the predicted probability and the observed target. The Receiver Operator Characteristic
is the plot of the true positive rate over the false positive rate.
```

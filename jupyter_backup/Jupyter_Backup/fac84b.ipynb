```python
def initials(name, name_part):
    names = name.split(' ')
    if name_part == "first":
        names[0] = names[0][0]
    elif name_part == "last":
        names[-1] = names[-1][0]
    else:
        for i in range(len(names)):
            if i != 0 and i != len(names) - 1:
                names[i] = names[i][0]
    return " ".join(names)
```

```python
print(initials("Ian Lewis Thornton Reinert", "first"))
print(initials("Ian Lewis Thornton Reinert", "last"))
print(initials("Ian Lewis Thornton Reinert", "middle"))
```

```python
def alter_name(name, add_char, name_part):
    names = name.split(' ')
    if name_part == "first":
        names[0] += add_char
    elif name_part == "last":
        names[-1] += add_char
    else:
        for i in range(len(names)):
            if i != 0 and i != len(names) - 1:
                names[i] += add_char
    return " ".join(names)
```

```python
print(alter_name("Ian Lewis Thornton Reinert", "e", "first"))
print(alter_name("Ian Lewis Thornton Reinert", "e", "last"))
print(alter_name("Ian Lewis Thornton Reinert", "e", "middle"))
```

```python
def homophone(name, name_part):
    names = name.split(' ')
    if name_part == "first":
        homophones = Search.perfectHomophones(names[0])
        if len(homophones) > 0:
            names[0] = homophones[0]
        else:
            names[0] = Search.closeHomophones(names[0])[0]
    elif name_part == "last":
        homophones = Search.perfectHomophones(names[-1])
        if len(homophones) > 0:
            names[-1] = homophones[0]
        else:
            names[-1] = Search.closeHomophones(names[0])[0]
    else:
        for i in range(len(names)):
            if i != 0 and i != len(names) - 1:
                homophones = Search.perfectHomophones(names[i])
                if len(homophones) > 0:
                    names[i] = homophones[0]
                else:
                    names[i] = Search.closeHomophones(names[0])[0]
                    
    return " ".join(names)
```

```python
print(homophone("Ian Thornton Reinert", "first"))
print(homophone("Jimmy Zhu", "first"))
print(homophone("Ian Thornton Reinert", "middle"))
```

```python
import random
def add_suffix(name):
    suffixes = ["Jr", "Sr", "I", "II", "III",]
    name += " " + random.choice(suffixes)
    
    return name 
```

```python
import sqlite3
import pandas as pd
import plotly.express as px
import os
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
import psycopg2
from sqlalchemy import create_engine
```

```python
# Connect to local db
engine = create_engine('postgresql://localhost/womens_directory_dump')
```

```python
# Query all category names and id per visit id (distinct takes out duplicate combos on same visit. Ex. [1,2] [1,2] -> [1,2])
query = """
        SELECT DISTINCT ahoy_visits.id AS visit_id, category_id, cat.name
        FROM ahoy_events
        JOIN ahoy_visits ON ahoy_visits.id = ahoy_events.visit_id
        JOIN categories AS cat ON cat.id = ahoy_events.category_id;
        """
visit_cats_df = pd.read_sql_query(query, engine)
visit_cats_df
```

```python
# Group all category_id's by visit_id's
# Returns each row is a list of the category id's per visit
# Group the visit_cats_df by visit id, returns 
by_visit_id = visit_cats_df.groupby('visit_id')
for visit_id in by_visit_id.groups:
    group = by_visit_id.get_group(visit_id)
    print(list(group['category_id']))
# get_group: returns a df from the group with the provided name
# Deprecated since version 2.1.0: The obj is deprecated and 
# will be removed in a future version. Do df.iloc[gb.indices.get(name)] instead of gb.get_group(name, obj=df).
```

```python
# Group category_id's per visit_id into series? Ex. [(8, 15)] category 8 and 15 were both visited by visit_id = 1
from itertools import combinations

by_visit = visit_cats_df.groupby('visit_id')
for visitor_id in by_visit.groups:
    group = by_visit.get_group(visitor_id)
    visited = list(group['category_id'])
    coms = combinations(visited, 2)
    print(visitor_id, list(coms))
# Output: Returns a list of list
# Transforms the list above to create unique combinations of each number in each list
# For example, it turns: [1, 3, 10, 16, 18] into: 
# [(1, 3), (1, 10), (1, 16), (1, 18), (3, 10), (3, 16), (3, 18), (10, 16), (10, 18), (16, 18)]
```

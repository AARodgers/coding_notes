```python
import sqlite3
import pandas as pd
import plotly.express as px
import os
```

```python
import psycopg2
from sqlalchemy import create_engine
```

```python
engine = create_engine('postgresql://localhost/womens_directory_dump')
```

```python
import pandas as pd
```

```python
query = """
        SELECT * FROM categories
        ORDER BY categories.id;
        """
cat_id_df = pd.read_sql_query(query, engine)
cat_id_df = cat_id_df[['id', 'name']]
cat_id_df
```

```python
# Download df to CSV file
# Get the path to the "Downloads" folder
downloads_folder = os.path.join(os.path.expanduser('~'), 'Downloads')


# Define the file name
file_name = 'wd_categories.csv'


# Combine the "Downloads" folder path with the file name
file_path = os.path.join(downloads_folder, file_name)


# df.to_csv
cat_id_df.to_csv(file_path, index=False)
print("CSV file has been created successfully in the Downloads folder.")
```

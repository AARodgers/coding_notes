#### What are the specific business objectives of this project? Define as precisely as possible.

Business objectives are defined by qualitatitve statements used to evaluate whether or not a company or entity is meeting the goals they are currently striving towards. At a high-level, the IEEE-CIS Fraud Detection overview, presented on Kaggle, aims to create a fraud detection system with a higher rate of efficacy than previous systems (Maximize fraud detection, minimize customer monetary loss). Given this scenario, measurable results directly related to IEEE-CIS's goals include:
1. Higher success rate's of fraud detection identified using a confusion matrix (TN, TP, FN, FN). Some of the key performance indicators tied to fraud detection success is included below. 
   - Accuracy (Total success in correctly identifying fraudulent and non-fraudulent cases)
   - ROC/AUC (Tug of war of optimizing Precision and Recall)
       - Precision - Of the fraudulent cases identified (wrong or right), how many were actually fraudulent.
         - How accurate is the system at correctly identifying someting as fraudulent
       - Recall - Of all of the ACTUAL fraudulent cases, how many were correctly identified.
         - Essentially, how often is the system able to identify something as fraudulent 
   - Fraud Rate - Cases that were identified as fraudulent over total process cased (as a ratio or percentage)
   - Incoming Pressure - Total volume of fraudulent transactions in transactional cases (used with approval rating and denial rating)
   - Approval Rating vs Denial Rating
       - How many transactions were approved vs how many transactions were denied
       - Determines approval rating for after screening and can be used in complement with incoming pressure to determine where approval rating should lie.
3. Robust system with extremely quick detection timing
   - System should have millisecond response and processing time for screening fraudulent transactions.
   - Streaming Data pipelines for processing & multiple data sources
4. Develop relevant machine learning models with up-to-date data in bi-monthly intervals (continuous adapatation)
5. Better customer experience in terms of minimizing monetary loss and time
   - Generate detailed and real-time reporting for users
   - Minimize loss for customers and cost for entity
6. Follow compliance and security guidelines to ensure security on sensitive data
   - E-commerce transactions contain large amounts of sensitve data and therefore should follow compliance and security guidelines.
7. Scalability and reliability
   - Systems just have up high up-time using failovers and redundancy as well should have the ability to handle large volumes of data

#### What data are available to pursue those objectives?

There are two datasets that been included in this project, identity and transaction. They are tabular datasets that have come in the form of .csv and have been split into training and test with a binary target of isFraud. They both contain a transaction ID column which allows for the joining of both datasets. The categorical features include ProductCD, cards1-6, addr1-2, R_P_emaildomain, and M1-9. 

1. Transaction Dataset - Contains 11 types of different features

    - TransactionAmt - amount of money spent in the transaction
    - ProductCD - product code for each transaction
    - Card1 - Card6 - payment information e.g. bank, card type, country, etc.
    - P_emaildomain - purchaser email domain
    - R_emaildomain - recipient email domain
    - M1 - M9 - names, address, card match
    - C1 - C14 - counting, e.g. how many addresses are associated with payment card
    - D1 - D15 - timedelta between transactions 
    - Addr - address 
    - dist - distance
    - Vxxx - features engineered by Vesta
    
2. Identity Dataset - Contains 3 types of different features (essentially just identity information)

    - DeviceType
    - DeviceInfo
    - id_12 - id_38

#### What are the success criteria for each of the project's objectives? Define as precisely as possible.

Specific success critiera for each of the project's objectives are generally case by case depending on things like resources, customer base, transactional volume, etc. But these are the general guidelines to follow:
1. Higher success rate of fraud detection.
   - Accuracy should at least be 90% when it comes to detecting fraud.
   - Precision and recall should both be relatively high, with precision being at least 90% and recall within the 70-90% range.
   - Fraud rate ideally should not be greater than 1%, but this is industry specific.
   - Approval rating and denial rating are dependent on incoming pressure.
     - Denial rating and incoming pressure should not be inversely proportional.
     - Approval rating should be around 80% while denial rating should be around 20% according to general industry guidelines.
        - If denial rate is too high, than there is a possibility of turning away good customers while high approval rate may result in more fraudulent transactions
3. Quick Detection Timing
   - Real-time detection and analyzation should be accomplished in milliseconds in order to fulfill the objective of "real-time" analytics
5. Relevant number of up-to-date machine learning models
   - Feature engineering should be applied to transform data into useful attributes
   - Number of models is subject to the situation, but each model should be tailored towards the specific business objective with high scores (f-1, precision, recall, accuracy, etc.)
   - models should be updated in consistent and frequent periods e.g. bi-monthly
7. Improved Customer Experience
   - Faster real-time reports to minimize monetary loss and lost time
        - can be measured through average time for charge back, money lost, number of denied chargebacks etc.
9. Following compliance and security guidelines
   -  Followed all compliance and security guidelines (AML, PSD2, etc.)
   -  Fulfilled all reporting duties and no penalties served
11. Scalability and reliablility
    - Reliable systems with high availability 99.999% (5 nines)
    - number of failsafes and replicas (for redundancy) to ensure this

#### Describe the data mining problem type (regression, classification, clustering, etc.)

This is a binary classification problem. The goal is to determine whether or not each of the transactions is fraudulent or non-fraudulent, which means there are only two possible options. The dataset included in the Kaggle prompt includes labeled data which points to supervised learning.

```python
import pandas as pd

# Sample data
data = {
    'sender_name': ['John Doe', 'Jane Smith', 'Alice Johnson', 'Bob Brown', 'Charlie Davis'],
    'receiver_name': ['Emily White', 'David Harris', 'Megan Clark', 'Chris Martin', 'Sophia Lee']
}

# Create the DataFrame
df = pd.DataFrame(data)

# Display the DataFrame
df
```

```python
import pandas as pd

def add_sender_first_name(df):
    """
    Add a new column called sender_first_name next to the sender_name column.
    The sender_first_name column contains the first name extracted from the sender_name column.
    """
    # Split the sender_name column into two parts
    name_split = df['sender_name'].str.split(n=1, expand=True)
    
    # Assign the first part to sender_first_name
    df['sender_first_name'] = name_split[0]
    
    # Reorder columns to place sender_first_name next to sender_name
    cols = df.columns.tolist()
    sender_name_index = cols.index('sender_name')
    cols.insert(sender_name_index + 1, cols.pop(cols.index('sender_first_name')))
    df = df[cols]
    
    return df
```

```python
df = add_sender_first_name(df)
```

```python
import pandas as pd

def add_sender_last_name(df):
    """
    Add a new column called sender_last_name next to the sender_name column.
    The sender_last_name column contains the last name extracted from the sender_name column.
    """
    # Split the sender_name column into parts
    name_split = df['sender_name'].str.split()
    
    # Assign the last part to sender_last_name
    df['sender_last_name'] = name_split.str[-1]
    
    # Reorder columns to place sender_last_name next to sender_name
    cols = df.columns.tolist()
    sender_name_index = cols.index('sender_name')
    cols.insert(sender_name_index + 2, cols.pop(cols.index('sender_last_name')))
    df = df[cols]
    
    return df
```

```python
df = add_sender_last_name(df)
df
```

# Zipping ranges into a dictionary

```python
# Create a list called keys from the range 899 to 948
keys = list(range(899, 949))
values = list(range(949,999))

# Merge the lists into a dictionary using zip
true_matches = dict(zip(keys, values))

# Need to manually add in these matches
true_matches[2] = 999
true_matches[3] = 1000

# Print the merged dictionary
print(true_matches)
```

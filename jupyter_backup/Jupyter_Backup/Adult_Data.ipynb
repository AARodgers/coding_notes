```python
import pandas as pd

# Assuming your data is in a list
data = ["State-gov, 77516, Bachelors, 13, Never-married, Adm-clerical, Not-in-family, White, Male, 2174, 0, 40, United-States, <=50K"]

# Create a DataFrame from the list
df = pd.DataFrame(data, columns=['original'])

# Split the 'original' column into multiple columns
split_columns = df['original'].str.split(',', expand=True)

# Optionally, you can assign column names
split_columns.columns = [
    'Workclass', 'Fnlwgt', 'Education', 'Education-Num', 'Marital-Status',
    'Occupation', 'Relationship', 'Race', 'Sex', 'Capital-Gain',
    'Capital-Loss', 'Hours-per-week', 'Native-Country', 'Income'
]

# Display the new DataFrame with separated columns
print(split_columns)
```

```python
# ============================================
# Bitcoin On Chain Data Analysis
# Using Blockchain.com Public API
# ============================================

import requests
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------------
# Helper function to pull chart data
# -------------------------------

def get_blockchain_chart(metric, timespan="180days"):
    """
    Pulls on chain data from Blockchain.com charts API
    """
    url = f"https://api.blockchain.info/charts/{metric}"
    params = {
        "timespan": timespan,
        "format": "json"
    }
    response = requests.get(url, params=params)
    data = response.json()
    return data["values"]

# -------------------------------
# Pull On Chain Metrics
# -------------------------------

# Number of transactions per day
tx_data = get_blockchain_chart("n-transactions")

# Hash rate (network security)
hashrate_data = get_blockchain_chart("hash-rate")

# Mining difficulty
difficulty_data = get_blockchain_chart("difficulty")

# -------------------------------
# Convert to DataFrames
# -------------------------------

tx_df = pd.DataFrame(tx_data)
hash_df = pd.DataFrame(hashrate_data)
diff_df = pd.DataFrame(difficulty_data)

# Convert timestamps
tx_df["date"] = pd.to_datetime(tx_df["x"], unit="s")
hash_df["date"] = pd.to_datetime(hash_df["x"], unit="s")
diff_df["date"] = pd.to_datetime(diff_df["x"], unit="s")

# Rename value columns
tx_df.rename(columns={"y": "transactions"}, inplace=True)
hash_df.rename(columns={"y": "hashrate"}, inplace=True)
diff_df.rename(columns={"y": "difficulty"}, inplace=True)

# Keep only needed columns
tx_df = tx_df[["date", "transactions"]]
hash_df = hash_df[["date", "hashrate"]]
diff_df = diff_df[["date", "difficulty"]]

# -------------------------------
# Merge On Chain Metrics
# -------------------------------

df = tx_df.merge(hash_df, on="date").merge(diff_df, on="date")
df = df.sort_values("date")

print("Sample on chain dataset")
print(df.head())

# ============================================
# Feature Engineering
# ============================================

# Rolling averages smooth noisy on chain data
df["tx_ma_7"] = df["transactions"].rolling(7).mean()
df["hashrate_ma_7"] = df["hashrate"].rolling(7).mean()

# Percentage change features
df["tx_change"] = df["transactions"].pct_change()
df["hashrate_change"] = df["hashrate"].pct_change()

# Drop NaN rows from rolling windows
df = df.dropna()

print("\nFeature engineered dataset")
print(df.head())

# ============================================
# VISUAL 1 Transactions Activity
# ============================================

plt.figure()
plt.plot(df["date"], df["transactions"])
plt.plot(df["date"], df["tx_ma_7"])
plt.title("Bitcoin Daily Transactions")
plt.xlabel("Date")
plt.ylabel("Transactions")
plt.show()

# Teaching point:
# Higher transaction counts indicate increased network usage

# ============================================
# VISUAL 2 Network Hashrate
# ============================================

plt.figure()
plt.plot(df["date"], df["hashrate"])
plt.plot(df["date"], df["hashrate_ma_7"])
plt.title("Bitcoin Network Hashrate")
plt.xlabel("Date")
plt.ylabel("Hashrate")
plt.show()

# Teaching point:
# Hashrate represents network security and miner confidence

# ============================================
# VISUAL 3 Difficulty Adjustment
# ============================================

plt.figure()
plt.plot(df["date"], df["difficulty"])
plt.title("Bitcoin Mining Difficulty")
plt.xlabel("Date")
plt.ylabel("Difficulty")
plt.show()

# Teaching point:
# Difficulty adjusts as miners enter or leave the network
```

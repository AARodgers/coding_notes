```python
import pandas as pd
import random
import numpy as np
from faker import Faker
```

```python
# Initialize Faker
fake = Faker()

# Seed for reproducibility
random.seed(42)
np.random.seed(42)
```

### Create fake GIIN's

Example GIIN: 123456.12345.12.123

See: https://www.irs.gov/businesses/corporations/fatca-registration-and-ffi-list-giin-composition-information

```python
# Create fake GIIN numbers
def generate_fake_giin():

    codes = ['LE', 'SL', 'ME', 'BR', 'SP', 'SF', 'SD', 'SS', 'SB']
    
    # Generate the FATCA ID (6 digits)
    fatca_id = ''.join([str(random.randint(0, 9)) for _ in range(6)])
    
    # Generate the Sponsoring Entity (5 digits)
    sponsoring_entity = ''.join([str(random.randint(0, 9)) for _ in range(5)])
    
    # Generate the Category Code (2 letters, ISO 3166-1 alpha-2 code)
    category_code = random.choice(codes)
    
    # Generate the Country Code (3 digits)
    country_code = ''.join([str(random.randint(0, 9)) for _ in range(3)])
    
    # Combine all parts to form the GIIN
    giin = f"{fatca_id}.{sponsoring_entity}.{category_code}.{country_code}"
    
    return giin

# Generate a fake GIIN
# fake_giin = generate_fake_giin()
# print(fake_giin)
```

### Create other fake PII

```python
# Create first name

def generate_random_first_name():
    first_name = fake.first_name()
    return first_name

# Generate a random first name
# random_first_name = generate_random_first_name()
# print(random_first_name)
```

```python
# Create fake phone number in format (xxx) xxx-xxxx
# Can change format of phone numbers later if needed for country codes and extensions

def generate_formatted_phone_number():
    phone_number = fake.numerify('(###) ###-####')
    return phone_number

# Generate a formatted fake phone number
formatted_phone_number = generate_formatted_phone_number()
print(formatted_phone_number)
```

### Create dataframe

```python
# Create a DataFrame with 1000 rows
data = {
    'first_name': [generate_random_first_name() for _ in range(1000)],
    'last_name': [generate_random_last_name() for _ in range(1000)],
    'phone_number': [generate_formatted_phone_number() for _ in range(1000)],
    'email': [generate_fake_email() for _ in range(1000)],
    'street_address': [generate_street_address() for _ in range(1000)],
    'city': [generate_single_word_city() for _ in range(1000)],
    'state': [generate_fake_state() for _ in range(1000)],
    'zip_code': [generate_fake_zip_code() for _ in range(1000)],
    'country': [tax_country() for _ in range(1000)],
    'fatca_id': [fatca_id() for _ in range(1000)], 
    'giin': [generate_fake_giin() for _ in range(1000)]
}

df = pd.DataFrame(data)

# Display the first few rows of the DataFrame
print(df.head())
```

```python
# Add a unique_id column ( per Splink requirements) and make it the first column
def add_unique_id(df):
    df.insert(0, 'unique_id', range(1, len(df) + 1))
    return df

# Add the unique_id column to the DataFrame
df = add_unique_id(df)

# Display the first few rows of the DataFrame
df.head(10)
```

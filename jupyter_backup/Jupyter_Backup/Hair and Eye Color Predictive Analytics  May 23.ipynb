```python
#Build a model
#Linear model
#Y ~ X

from statsmodels.formula.api import ols
model = ols('Freq ~ Sex + Hair + Eye', data = my_data)
fitted_model = model.fit()
fitted_model.summary()
```

```python
# group data by gender and hair color
hair_color = my_data.groupby(['Sex','Hair'])['Freq'].sum()
hair_color

# # transform data frame
hair_color_df = hair_color.unstack(level=0)

# # print out the data frame
hair_color_df
# create a copy of existing data frame so that we can add total row and column without making changes to the df
hair_color_df_with_totals = hair_color_df.copy()
hair_color_df_with_totals
# add total column
hair_color_df_with_totals['Total Col'] = hair_color_df_with_totals.sum(axis=1)

# # add total row
hair_color_df_with_totals.loc['Total Row'] = hair_color_df_with_totals.sum(axis=0)

# # print out the data frame
hair_color_df_with_totals

# below will be a 4 x 2 contigency table
```

```python
# exctract total values for calculating expected values
row_mult = hair_color_df_with_totals.iloc[4:,:-1].values
row_mult

col_mult = hair_color_df_with_totals.iloc[:-1,2:].values
col_mult

# # exctract grand total
total_general = hair_color_df_with_totals.iloc[4:,-1:].values
total_general
```

```python
# create an array of expected values
expected = (row_mult*col_mult/total_general).round(2)

# print array out
expected
```

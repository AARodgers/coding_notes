```python
# Function to extract street number and name for business_mailing_address column
def extract_number_and_street(full_address):
    """
    This function takes a full address and extracts the number and street name.
    """
    # Split the address by newline character
    address_parts = full_address.split('\n')
    
    # The first part contains the number and street name
    number_and_street = address_parts[0]
    
    return number_and_street
```

```python
# Apply the function to the column
f8957_df['business_mailing_address'] = f8957_df['business_mailing_address'].apply(extract_number_and_street)
```

```python
# Function to remove periods from an address
def remove_periods(address):
    """
    This function removes periods from the address.
    """
    return address.replace('.', '')
```

```python
# Apply the function to remove periods from business_mailing_address column
f8957_df['business_mailing_address'] = f8957_df['business_mailing_address'].apply(remove_periods)
```

```python
# Function to replace abbreviations with full words per Slink data prep requirements
def replace_abbreviations(address):
    """
    This function replaces common abbreviations in the address.
    """
    abbreviations = {
        'Apt': 'Apartment',
        'St': 'Street',
        'Ave': 'Avenue',
        'Rd': 'Road',
        'Blvd': 'Boulevard',
        'Dr': 'Drive',
        'Ln': 'Lane',
        'Ct': 'Court',
        'Pl': 'Place',
        'Sq': 'Square',
        'Terr': 'Terrace',
        'Pkwy': 'Parkway',
        'Cir': 'Circle'
    }
    
    for abbr, full in abbreviations.items():
        address = address.replace(abbr, full)
    
    return address
```

```python
# Apply the function to replace abbreviations in business_mailing_address column
f8957_df['business_mailing_address'] = f8957_df['business_mailing_address'].apply(replace_abbreviations)
```

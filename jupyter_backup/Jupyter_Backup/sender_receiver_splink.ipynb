```python
import pandas as pd
import numpy as np

def standardize_missing_values(df):
    """
    Standardize cells containing 'nan', 'Nan', 'None', or other representations of missing values to true null values (NaN).
    """
    # Replace 'nan', 'Nan', 'None', and other representations with np.nan
    df.replace(['nan', 'Nan', 'None', 'xx', None], np.nan, inplace=True)
    
    return df

df = standardize_missing_values(df)
```

```python
def add_sender_first_name(df):
    """
    Add a new column called sender_first_name next to the sender_name column.
    The sender_first_name column contains the first name extracted from the sender_name column.
    """
    # Split the sender_name column into two parts
    name_split = df['sender_name'].str.split(n=1, expand=True)
    
    # Assign the first part to sender_first_name
    df['sender_first_name'] = name_split[0]
    
    # Reorder columns to place sender_first_name next to sender_name
    cols = df.columns.tolist()
    sender_name_index = cols.index('sender_name')
    cols.insert(sender_name_index + 1, cols.pop(cols.index('sender_first_name')))
    df = df[cols]
    
    return df
```

```python
df = add_sender_first_name(df)
df
```

```python
def add_sender_last_name(df):
    """
    Add a new column called sender_last_name next to the sender_name column.
    The sender_last_name column contains the last name extracted from the sender_name column.
    """
    # Split the sender_name column into parts
    name_split = df['sender_name'].str.split()
    
    # Assign the last part to sender_last_name
    df['sender_last_name'] = name_split.str[-1]
    
    # Reorder columns to place sender_last_name next to sender_name
    cols = df.columns.tolist()
    sender_name_index = cols.index('sender_name')
    cols.insert(sender_name_index + 2, cols.pop(cols.index('sender_last_name')))
    df = df[cols]
    
    return df
```

Next steps: 
- Did above: blocking rules to generate pairwise record comparisons
- Train linkage model 
- Estimate a probabilistic linkage model to score each of the comparisons above ( first initial and last name)
    - to learn the relative importance of different parts of your data for the purpose of data linking
    - example: dob is stronger indicator that two rows are same entity than gender
- The relative importance of different information is captured in the (partial) 'match weights', which can be learned from your data.
- The match weights are are derived from the m and u parameters of the underlying Fellegi Sunter model.
- The concept of a Comparison has a specific definition within Splink: it defines how data from one or more input columns is compared.
    - For example, one Comparison may represent how similarity is assessed for a person's date of birth.

```python
# Example of a generic Splink comparison function using fuzzy matching and levenshtein distance

import splink.comparison_library as cl

city_comparison = cl.LevenshteinAtThresholds("city", 2)
print(city_comparison.get_comparison("duckdb").human_readable_description)
```

```python
# Step: Specify linkage model with your own comparisons and blocking rules
# Comparisons are specified as part of the Splink settings, a Python dictionary which controls all of the 
# configuration of a Splink model:
from splink import Linker, SettingsCreator, block_on, DuckDBAPI

settings = SettingsCreator(
    link_type="dedupe_only",
    comparisons=[
        cl.NameComparison("sender_first_name"),
        cl.NameComparison("sender_last_name"),
        cl.LevenshteinAtThresholds("sender_address_full", 1),
        cl.ExactMatch("sender_last_name").configure(term_frequency_adjustments=True)
    ],
    # will only check for duplicates amongst records where either the first name first intial AND last name or lastname is identical
    blocking_rules_to_generate_predictions=[
        block_on("substr(sender_first_name, 1, 2)", "sender_last_name"),
        block_on("sender_last_name"),

    ],
    retain_intermediate_calculation_columns=True,
)

linker = Linker(df, settings, db_api=DuckDBAPI())
```

```python
# Estimate the probability_two_random_records_match
# The u values are the proportion of records falling into each ComparisonLevel amongst truly non-matching records.
# The m values are the proportion of records falling into each ComparisonLevel amongst truly matching records
# TO DO: need to look more into how to determine this estimation: https://github.com/moj-analytical-services/splink/issues/462#issuecomment-1227027995
# Whatever percentage you set the recall to you are saying that the deterministic_rules you set should cover about that percentage of true matches

deterministic_rules = [
    block_on("substr(sender_first_name, 1, 2)"),
    block_on("sender_last_name")
]

linker.training.estimate_probability_two_random_records_match(deterministic_rules, recall=0.2)
```

```python
# Estimate u parameter
linker.training.estimate_u_using_random_sampling(max_pairs=1e6)
```

Next Step: Load the estimated model and use it to make predictions of which pairwise record comparisons match

```python
from splink import Linker, DuckDBAPI, splink_datasets

import pandas as pd

pd.options.display.max_columns = 1000

db_api = DuckDBAPI()
df = df
```

```python
# Load estimated model
import json

file_path = "/Users/amandarodgers/code/splink_practice/saved_model_from_demo.json"

# Open the file and load the settings
with open(file_path, 'r') as f:
    settings = json.load(f)

linker = Linker(df, settings, db_api=DuckDBAPI())
```

```python
# Run the model which will: 
# Generate all pairwise record comparisons that match at least one of the blocking_rules_to_generate_predictions
# Use the rules specified in the Comparisons to evaluate the similarity of the input data
# Use the estimated match weights, applying term frequency adjustments where requested to produce the final match_weight and match_probability scores
# Optionally, a threshold_match_probability or threshold_match_weight can be provided, which will drop any row where the predicted score is below the threshold.

df_predictions = linker.inference.predict(threshold_match_probability=0.2)
predictions_df = df_predictions.as_pandas_dataframe(limit=1000)
```

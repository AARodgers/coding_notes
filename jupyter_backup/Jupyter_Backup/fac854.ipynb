````python
import pandas as pd
import random

# Generate random values for each column
principals = [random.randint(5000, 100000) for _ in range(100)]
inflations = [round(random.uniform(0.02, 0.09), 2) for _ in range(100)]
interests = [principal * inflation for principal, inflation in zip(principals, inflations)]

# Introduce incorrect interest values for 10% of the entries
num_wrong_entries = int(len(interests) * 0.1)
indices_to_change = random.sample(range(len(interests)), num_wrong_entries)
for idx in indices_to_change:
    interests[idx] = random.randint(5000, 100000)  # Assigning a random value

# Create DataFrame
data = {'Principal': principals, 'Inflation': inflations, 'Interest': interests}
df = pd.DataFrame(data)
df
````

````python
help(df.drop)
````

````python
sub_set_df = df.drop(columns=['Interest'])
sub_set_df
````

````python
correct_interest = sub_set_df.Principal * sub_set_df.Inflation
correct_interest
````

````python
# Name it CorrectInterest
correct_interest.name = 'CorrectInterest'
correct_interest
````

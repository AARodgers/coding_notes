```python
import pandas as pd
import random
import numpy as np
from faker import Faker
```

```python
# Load unique_people_transactions.csv ( 1000 unique people and transactions)
df = pd.read_csv('datasets/dataset_1.csv')
```

```python
df.head(5)
```

```python
# Make dataset_2 by modifying the first 100 rows of dataset_1 and changing the first_name column to the first character in the string
def modify_first_name_by_unique_id(df):
    """
    Modify the rows with unique_id from 1 to 100 in the DataFrame, changing the `first_name` column to contain only the first character
    of the original `first_name`.

    Parameters:
    df (pd.DataFrame): The input DataFrame with columns `first_name` and `unique_id`.

    Returns:
    pd.DataFrame: The modified DataFrame.
    """
    # Filter the DataFrame to get rows with unique_id from 1 to 100
    rows_to_modify = df[df['unique_id'].between(1, 100)].index
    
    # Modify the `first_name` column for the selected rows
    df.loc[rows_to_modify, 'first_name'] = df.loc[rows_to_modify, 'first_name'].str[0]
    
    return df

# Modify the DataFrame
dataset_2 = modify_first_name_by_unique_id(df)
dataset_2.head(105) # Print the top 105 rows to see the changes
```

```python
# Convert dataset_2 to a csv file
import os

def save_dataframe_to_csv(df, filename='dataset_2.csv', folder='datasets'):
    """
    Save the DataFrame as a CSV file in the specified folder.

    Parameters:
    df (pd.DataFrame): The DataFrame to save.
    filename (str): The name of the CSV file. Default is 'dataset_2.csv'.
    folder (str): The folder where the file will be saved. Default is 'datasets'.

    Returns:
    str: The path to the saved file.
    """
    # Ensure the folder exists
    if not os.path.exists(folder):
        os.makedirs(folder)
    
    # Create the full file path
    file_path = os.path.join(folder, filename)
    
    # Save the DataFrame to a CSV file
    df.to_csv(file_path, index=False)
    
    return file_path

# Save the DataFrame as a CSV file
file_path = save_dataframe_to_csv(dataset_2)
print(f"DataFrame saved to: {file_path}")
```

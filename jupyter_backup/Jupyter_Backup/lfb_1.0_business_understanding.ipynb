# Business Understanding
During this stage, you should take some time to understand the questions presented by the data from a busines perspective. Many of these are very general questions and they may not be as important or applicable as some other stages of this project for training purposes, but it's good practice to begin any project with a general understanding of what problems you are trying to solve.

Consider the following questions and answer to the best of your ability, based on the data and project documentation.

#### What are the specific business objectives of this project? Define as precisely as possible.

The business objective of this project is to create a model that can predict fraudulent credit card transactions. The sponsor of the Kaggle competition uses machine learning models to flag potentially fraudulent transactions. These flagged transactions can be used to freeze credit card accounts and send out verification alerts to the credit card user. Creating improved models, then, will help the sponsor better identify suspicious activity, reducing both fraud and the false positives that inconvenience consumers.

#### What data are available to pursue those objectives?

The Kaggle competition makes data available from Vesta Corporation, which is a company that analyzes and approves credit card transactions. However, due to privacy concerns much of the data is masked to a degree. There are slightly over 400 variables provided, but most of these are not identified. Identified variables include card type (credit vs. debit), card brand, address, email domain, and distance. The training data has ~590k rows. 

#### What resources? (programs, libraries, etc.)

We will likely use Python's data science libraries for this project, including pandas, numpy, and sklearn. For data visualizations we can use matplotlib and seaborn. 

#### What are the success criteria for each of the project's objectives? Define as precisely as possible.

The Kaggle competition defines success as the model with the highest roc-auc score. Roc-auc score measures the area under a curve that shows the True Positive and False positive rates at all classification thresholds. 

#### Describe the data mining problem type (regression, classification, clustering, etc.)

This is a classification problem as the goal is to predict which transactions are fraudulent, and there are only two possible options (fraudulent and not fraudulent). 

#### What are the specific technical goals for the project?¶

The technical goal of the project is to develop a model that can create predictions for submission to the Kaggle project.

```python
import pandas as pd
```

```python
# Load the dataset_1.csv
dataset_1 = pd.read_csv('datasets/dataset_1.csv')

# Display the first few rows of the DataFrame to verify that the data has been loaded correctly
dataset_1.head(5)
```

```python
# Load the dataset_2.csv
dataset_2 = pd.read_csv('datasets/dataset_2.csv')

# Display the first few rows of the DataFrame to verify that the data has been loaded correctly
dataset_2.head(5)
```

```python
# Explore amount of missing data because columns with higher levels of null values are less useful for data linking
# Missing data for dataset_1
from splink.exploratory import completeness_chart
from splink import DuckDBAPI
db_api = DuckDBAPI()
completeness_chart(dataset_1, db_api=db_api)
```

```python
# Explore amount of missing data because columns with higher levels of null values are less useful for data linking
# Missing data for dataset_2
from splink.exploratory import completeness_chart
from splink import DuckDBAPI
db_api = DuckDBAPI()
completeness_chart(dataset_2, db_api=db_api)
```

## Analyze distribution of data:

```python
# Columns with higher cardinality (number of distinct values) are usually more useful for data linking
# Skew of values are important. A column that has a majority of same values ( ex. cities is 75% London) 
# is less useful for linkage than if citites were equally distributed
```

```python
# Analyze distribution of distinct values in each column for dataset_1
from splink.exploratory import profile_columns
profile_columns(dataset_1, db_api=DuckDBAPI(), top_n=10, bottom_n=5)
```

```python
# Analyze distribution of distinct values in each column for dataset_2
from splink.exploratory import profile_columns
profile_columns(dataset_2, db_api=DuckDBAPI(), top_n=10, bottom_n=5)
```
